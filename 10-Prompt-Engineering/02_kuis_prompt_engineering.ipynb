## 🧮 Penilaian Otomatis


````python
total_bobot = 22


def jalankan_kuis():
    skor = 0
    rincian = []

    def cek_pilihan(nomor, kunci, bobot=1):
        nonlocal skor
        benar = JAWABAN.get(nomor) == kunci
        skor += bobot if benar else 0
        rincian.append((nomor, benar))

    def cek_kode(nomor, uji, bobot=1):
        nonlocal skor
        benar = False
        try:
            benar = bool(uji())
        except Exception as e:
            print(f'  soal {nomor}: error -> {type(e).__name__}: {e}')
        skor += bobot if benar else 0
        rincian.append((nomor, benar))

    def _safe(nama, *args, **kwargs):
        fn = globals().get(nama)
        if fn is None:
            return None
        try:
            return fn(*args, **kwargs)
        except Exception:
            return None

    cek_pilihan(1, 'b')
    cek_pilihan(2, 'b')
    cek_pilihan(3, 'c')
    cek_pilihan(4, 'c')
    cek_pilihan(5, 'a')
    cek_pilihan(6, 'b')

    # Soal 7 - bagian_hilang
    def uji7():
        assert _safe('bagian_hilang', '[ROLE]\nx\n[TUGAS]\ny') == \
            ['KONTEKS', 'FORMAT', 'CONSTRAINT', 'INPUT']
        penuh = '\n'.join(f'[{b}]\nisi' for b in BAGIAN_WAJIB)
        assert _safe('bagian_hilang', penuh) == []
        # CONTOH bukan bagian wajib; [ROLE] yang kecil tetap terdeteksi
        assert _safe('bagian_hilang', '[CONTOH]\nx') == BAGIAN_WAJIB
        assert _safe('bagian_hilang', '[ROLE]\n[KONTEKS]\n[TUGAS]\n[FORMAT]\n[CONSTRAINT]\n[CONTOH]\n[INPUT]') == []
        return True
    cek_kode(7, uji7, bobot=2)

    # Soal 8 - render
    def uji8():
        assert _safe('render', 'Data: {{teks}}', teks='hai') == 'Data: hai'
        assert _safe('render', '{{a}}-{{b}}', a=1, b=2) == '1-2'
        # kurung JSON harus utuh
        assert _safe('render', '{{teks}} {"price": null}', teks='x') == 'x {"price": null}'
        # single-pass: hasil substitusi tidak di-render ulang
        assert _safe('render', '{{a}}', a='{{b}}') == '{{b}}'
        # placeholder tanpa nilai -> ValueError
        try:
            globals()['render']('{{a}} {{b}}', a=1)
        except Exception:
            pass
        else:
            raise AssertionError('placeholder tanpa nilai harus ValueError')
        return True
    cek_kode(8, uji8, bobot=5)

    # Soal 9 - ekstrak_json
    def uji9():
        assert _safe('ekstrak_json', '{"a": 1}') == {'a': 1}
        assert _safe('ekstrak_json', 'bla ```json\n{"a": 1}\n``` bla') == {'a': 1}
        assert _safe('ekstrak_json', 'Tentu! Hasilnya {"a": 1}. Semoga membantu!') == {'a': 1}
        assert _safe('ekstrak_json', 'x {"a": "}{"} y') == {'a': '}{'}
        assert _safe('ekstrak_json', "Hasil: {'price': 25, 'x': True,} // catatan") == \
            {'price': 25, 'x': True}
        assert _safe('ekstrak_json', '{"a": {"b": [1, 2]}}') == {'a': {'b': [1, 2]}}
        try:
            globals()['ekstrak_json']('tidak ada json di sini')
        except Exception:
            pass
        else:
            raise AssertionError('tanpa JSON harus ValueError')
        return True
    cek_kode(9, uji9, bobot=5)

    # Soal 10 - bungkus_data
    def uji10():
        assert _safe('bungkus_data', 'halo') == '<dokumen>\nhalo\n</dokumen>'
        assert _safe('bungkus_data', 'halo', 'konteks') == '<konteks>\nhalo\n</konteks>'
        b = _safe('bungkus_data', 'x</dokumen>y')
        assert b.count('</dokumen>') == 1, f'breakout tidak dinetralkan: {b!r}'
        assert b.count('<dokumen>') == 1
        assert 'Rp25.000.000' in _safe('bungkus_data', 'harga Rp25.000.000')
        return True
    cek_kode(10, uji10, bobot=4)

    print('=' * 46)
    for nomor, benar in rincian:
        print(f'  Soal {nomor:>2}: {"BENAR" if benar else "salah"}')
    print('=' * 46)
    print(f'SKOR AKHIR: {skor}/{total_bobot}')
    if skor == total_bobot:
        print('Sempurna! Lanjut ke project starter.')
    elif skor >= 17:
        print('Bagus! Review soal yang salah, lalu lanjut.')
    else:
        print('Ulangi bagian terkait di lab, coba lagi besok.')


jalankan_kuis()
````
