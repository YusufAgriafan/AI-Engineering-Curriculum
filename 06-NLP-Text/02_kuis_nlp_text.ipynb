## 🧮 Penilaian Otomatis

`````python
import numpy as np

def sigmoid(z):
    return 1.0 / (1.0 + np.exp(-np.clip(z, -30, 30)))

total_bobot = 23


def jalankan_kuis():
    skor = 0
    rincian = []

    def cek_pilihan(nomor, kunci, bobot=1):
        nonlocal skor
        benar = JAWABAN.get(nomor) == kunci
        skor += bobot if benar else 0
        rincian.append((nomor, benar))

    def cek_kode(nomor, uji, bobot=1):
        nonlocal skor
        benar = False
        try:
            benar = bool(uji())
        except Exception as e:
            print(f'  soal {nomor}: error -> {type(e).__name__}: {e}')
        skor += bobot if benar else 0
        rincian.append((nomor, benar))

    def approx(a, b, tol=1e-9):
        return np.allclose(np.asarray(a, dtype=float), np.asarray(b, dtype=float), atol=tol)

    def _safe(nama, *args):
        # fungsi soal yang belum diisi tidak boleh menjatuhkan seluruh penilaian
        fn = globals().get(nama)
        if fn is None:
            return None
        try:
            return fn(*args)
        except Exception:
            return None

    cek_pilihan(1, 'a')
    cek_pilihan(2, 'b')
    cek_pilihan(3, 'c')
    cek_pilihan(4, 'c')

    cek_kode(5, lambda: isinstance(pad_sequences([[5, 2], [3, 9, 7, 4]], 3), np.ndarray)
                      and pad_sequences([[5, 2], [3, 9, 7, 4]], 3).tolist() == [[5, 2, 0], [3, 9, 7]]
                      and pad_sequences([[7], [1, 2]], 4).tolist() == [[7, 0, 0, 0], [1, 2, 0, 0]]
                      and pad_sequences([], 2).shape == (0, 2), bobot=3)

    _w2i = {'a': 0, 'b': 1, 'c': 2}
    cek_kode(6, lambda: sorted(pasangan_skipgram([['a', 'b', 'c']], _w2i, window=1))
                      == [(0, 1), (1, 0), (1, 2), (2, 1)]
                      and sorted(pasangan_skipgram([['a', 'b']], _w2i, window=1))
                      == [(0, 1), (1, 0)]
                      and sorted(pasangan_skipgram([['a', 'b', 'c']], _w2i, window=2))
                      == [(0, 1), (0, 2), (1, 0), (1, 2), (2, 0), (2, 1)], bobot=2)

    v0 = np.array([0.5, -0.1])
    u0 = np.array([0.2, 0.8])
    s0 = sigmoid(float(v0 @ u0))
    cek_kode(7, lambda: approx(sg_step(v0.copy(), u0.copy(), 0.0)[0], -np.log(s0))
                      and approx(sg_step(v0.copy(), u0.copy(), 0.0)[1], v0)
                      and approx(sg_step(v0.copy(), u0.copy(), 0.0)[2], u0)
                      and (sg_step(*_safe('sg_step', v0.copy(), u0.copy(), 0.5)[1:], 0.0)[0]
                           < sg_step(v0.copy(), u0.copy(), 0.0)[0])
                      and approx(sg_step(v0.copy(), u0.copy(), 0.5)[1], v0 - 0.5 * (s0 - 1) * u0)
                      and approx(sg_step(v0.copy(), u0.copy(), 0.5)[2], u0 - 0.5 * (s0 - 1) * v0), bobot=4)

    cek_kode(8, lambda: bigram_matrix([0, 1, 0, 2], 3, alpha=0.5).shape == (3, 3)
                      and approx(bigram_matrix([0, 1, 0, 2], 3, alpha=0.5).sum(axis=1), 1.0)
                      and approx(bigram_matrix([0, 1, 0, 2], 3, alpha=0.5)[0, 1], 2.5 / 6.5)
                      and approx(bigram_matrix([0, 1, 0, 2], 3, alpha=0.5)[0, 0], 1.5 / 6.5)
                      and approx(bigram_matrix([], 3, alpha=0.0).sum(axis=1), 1.0), bobot=3)

    logits_s = np.array([3.0, 2.0, 1.0, 0.0])
    rng9a = np.random.default_rng(11)
    sampel_Tk = [sample_with_temp(logits_s, 0.2, rng9a) for _ in range(200)]
    rng9b = np.random.default_rng(12)
    sampel_Tb = [sample_with_temp(logits_s, 2.5, rng9b) for _ in range(200)]
    freq = lambda s, i: sum(1 for x in s if x == i) / len(s)
    cek_kode(9, lambda: freq(sampel_Tk, 0) > 0.95
                      and freq(sampel_Tb, 1) > freq(sampel_Tb, 2) > 0.05
                      and all(0 <= int(x) < 4 for x in sampel_Tb + sampel_Tk), bobot=3)

    r = np.random.default_rng(0)
    x_t = r.normal(size=(2, 5))
    h_prev = r.normal(size=(2, 3))
    Wx = r.normal(size=(5, 3))
    Wh = r.normal(size=(3, 3))
    bh = r.normal(size=(3,)) * 0.1
    hasil = _safe('rnn_step_forward', x_t, h_prev, Wx, Wh, bh)
    h_ref = np.tanh(x_t @ Wx + h_prev @ Wh.T + bh)
    cek_kode(10, lambda: hasil is not None and len(hasil) == 2
                       and approx(hasil[0], h_ref)
                       and approx(rnn_step_backward(np.ones_like(h_ref), hasil[1]),
                                  1 - h_ref ** 2)
                       and approx(rnn_step_backward(h_ref, hasil[1]),
                                  h_ref * (1 - h_ref ** 2)), bobot=4)

    print('=' * 46)
    for nomor, benar in rincian:
        print(f'  Soal {nomor:>2}: {"BENAR" if benar else "salah"}')
    print('=' * 46)
    print(f'SKOR AKHIR: {skor}/{total_bobot}')
    if skor == total_bobot:
        print('Sempurna! Lanjut ke project starter.')
    elif skor >= 18:
        print('Bagus! Review soal yang salah, lalu lanjut.')
    else:
        print('Ulangi bagian terkait di lab, coba lagi besok.')

jalankan_kuis()
`````
