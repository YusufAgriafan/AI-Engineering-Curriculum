### Soal 9 - `rmse_mae`Dua metrik satu error vector: hitung `err` sekali, turunkan keduanya.Ingat: RMSE ≥ MAE selalu (Cauchy-Schwarz), dan RMSE menghukum outlier lebih keras.

````python
import numpy as np

def rmse_mae(y_true, y_pred):
    err = np.asarray(y_true, dtype=float) - np.asarray(y_pred, dtype=float)
    rmse = float(np.sqrt(np.mean(err ** 2)))
    mae = float(np.mean(np.abs(err)))
    return rmse, mae

r, m = rmse_mae(np.array([1., 2.]), np.array([2., 4.]))
print(r, m)   # 1.5811..., 1.5
r, m = rmse_mae(np.array([1., 9.]), np.array([0., 0.]))
print(r, m)   # 6.40 vs 5.0 - RMSE menghukum outlier
````

### Soal 10 - `confusion`Konvensi `>=` wajib - tanpa itu, kasus skor tepat di threshold ambigu dan duaorang menghitung dua angka berbeda dari model yang sama.

````python
def confusion(y_true01, y_score, thr):
    y_true01 = np.asarray(y_true01).astype(int)
    y_score = np.asarray(y_score, dtype=float)
    pred = (y_score >= thr).astype(int)
    return {'tp': int(np.sum((y_true01 == 1) & (pred == 1))),
            'fp': int(np.sum((y_true01 == 0) & (pred == 1))),
            'fn': int(np.sum((y_true01 == 1) & (pred == 0))),
            'tn': int(np.sum((y_true01 == 0) & (pred == 0)))}

print(confusion([1, 0, 1, 0], [0.9, 0.8, 0.3, 0.1], 0.5))
print(confusion([0], [0.5], 0.5))   # {'fp': 1} - tepat di thr = positif
````

### Soal 11 - `prf`Konvensi produksi: pembagi nol → 0.0, JANGAN crash. Model yang tak pernahprediksi positif memang layak dapat F1=0 - bukan exception di tengah serving.F1 = harmonic mean: hanya tinggi kalau P dan R KEDUANYA tinggi.

````python
def prf(cm):
    tp, fp, fn = cm['tp'], cm['fp'], cm['fn']
    p = tp / (tp + fp) if (tp + fp) else 0.0
    r = tp / (tp + fn) if (tp + fn) else 0.0
    f1 = 2 * p * r / (p + r) if (p + r) else 0.0
    return {'precision': p, 'recall': r, 'f1': f1}

print(prf({'tp': 3, 'fp': 1, 'fn': 2, 'tn': 4}))   # {'precision': 0.75, 'recall': 0.6, 'f1': 0.666...}
print(prf({'tp': 0, 'fp': 0, 'fn': 5, 'tn': 5}))   # semua 0 - tanpa crash
````

### Soal 12 - `ece_bin`Dua jebakan: (1) `min(int(p·n_bins), n_bins−1)` agar p=1.0 masuk bin terakhir,(2) bin kosong dilewati (bukan membagi nol). Contoh kuis: bins 1, 5, 8, 9 masing-masing 1 sampel → ECE = 0.25·(|0−0.1| + |0−0.2| + |1−0.8| + |1−0.9|) = **0.15**.

````python
def ece_bin(y_true01, p, n_bins=10):
    y_true01 = np.asarray(y_true01).astype(int)
    p = np.asarray(p, dtype=float)
    idx = np.minimum((p * n_bins).astype(int), n_bins - 1)
    n = len(p)
    total = 0.0
    for b in range(n_bins):
        mask = idx == b
        nb = int(mask.sum())
        if nb == 0:
            continue
        conf = float(p[mask].mean())
        acc = float(y_true01[mask].mean())
        total += (nb / n) * abs(acc - conf)
    return float(total)

print(ece_bin(np.array([1, 0, 1, 0]), np.array([0.9, 0.1, 0.8, 0.2]), 10))  # 0.15
print(ece_bin(np.array([1, 0]), np.array([1.0, 0.0]), 10))  # 0.0 - sempurna
````
