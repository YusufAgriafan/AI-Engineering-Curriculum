## 🧮 Penilaian Otomatis

```python
import numpy as np

total_bobot = 23


def jalankan_kuis():
    skor = 0
    rincian = []

    def cek_pilihan(nomor, kunci, bobot=1):
        nonlocal skor
        benar = JAWABAN.get(nomor) == kunci
        skor += bobot if benar else 0
        rincian.append((nomor, benar))

    def cek_kode(nomor, uji, bobot=1):
        nonlocal skor
        benar = False
        try:
            benar = bool(uji())
        except Exception as e:
            print(f'  soal {nomor}: error -> {type(e).__name__}: {e}')
        skor += bobot if benar else 0
        rincian.append((nomor, benar))

    def approx(a, b, tol=1e-9):
        return np.allclose(np.asarray(a, dtype=float), np.asarray(b, dtype=float), atol=tol)

    def _safe(nama, *args):
        # fungsi soal yang belum diisi tidak boleh menjatuhkan seluruh penilaian
        fn = globals().get(nama)
        if fn is None:
            return None
        try:
            return fn(*args)
        except Exception:
            return None

    cek_pilihan(1, 'b')
    cek_pilihan(2, 'b')
    cek_pilihan(3, 'c')
    cek_pilihan(4, 'b')

    img = np.arange(9, dtype=float).reshape(3, 3)
    cek_kode(5, lambda: conv_valid(img, np.ones((2, 2))).shape == (2, 2)
                      and approx(conv_valid(img, np.ones((2, 2))), [[8., 12.], [20., 24.]])
                      and approx(conv_valid(img, np.eye(2)), [[4., 6.], [10., 12.]])
                      and approx(conv_valid(np.ones((4, 4)), np.ones((3, 3))), 9 * np.ones((2, 2))), bobot=3)

    cek_kode(6, lambda: ukuran_output(32, 3) == 30
                      and ukuran_output(32, 3, 1) == 32
                      and ukuran_output(32, 3, 0, 2) == 15
                      and ukuran_output(28, 3, 1, 2) == 14
                      and ukuran_output(8, 5) == 4, bobot=2)

    x = np.arange(16, dtype=float).reshape(4, 4)
    cek_kode(7, lambda: maxpool2x2(x).shape == (2, 2)
                      and approx(maxpool2x2(x), [[5., 7.], [13., 15.]])
                      and maxpool2x2(np.ones((2, 2)))[0, 0] == 1.0, bobot=3)

    dOut = np.array([[1., 2.], [3., 4.]])
    dX = _safe('maxpool2x2_backward', dOut, x)
    xseri = np.ones((2, 2))
    cek_kode(8, lambda: dX.shape == (4, 4)
                      and approx(dX.sum(), 10.0)
                      and dX[1, 1] == 1.0 and dX[0, 0] == 0.0
                      and dX[1, 3] == 2.0 and dX[3, 1] == 3.0 and dX[3, 3] == 4.0
                      and approx(maxpool2x2_backward(np.ones((2, 2)), xseri).sum(), 1.0), bobot=4)

    g = np.arange(16, dtype=float).reshape(4, 4)
    g0 = g.copy()
    ok9 = False
    try:
        out9 = shift_zero(g, 1, 0)
        ok9 = (out9.shape == (4, 4) and np.allclose(g, g0)
               and approx(out9[1:], g[:3]) and approx(out9[0], np.zeros(4))
               and approx(shift_zero(g, 0, -1)[:, :-1], g[:, 1:])
               and approx(shift_zero(g, 0, -1)[:, -1], np.zeros(4))
               and approx(shift_zero(g, 0, 0), g))
    except Exception:
        ok9 = False
    cek_kode(9, lambda: ok9, bobot=3)

    img10 = np.zeros((5, 5)); img10[:, 3:] = 1.0
    K10 = _safe('filter_tepi_v')
    S10 = _safe('tepi_v_score', img10)
    kref = np.array([[-1., 0., 1.]] * 3)
    cek_kode(10, lambda: np.asarray(K10).shape == (3, 3)
                       and approx(K10, kref)
                       and S10.shape == (3, 3)
                       and approx(S10[:, 1:3], 3.0 * np.ones((3, 2)))
                       and approx(S10[:, 0], 0.0), bobot=4)

    print('=' * 46)
    for nomor, benar in rincian:
        print(f'  Soal {nomor:>2}: {"BENAR" if benar else "salah"}')
    print('=' * 46)
    print(f'SKOR AKHIR: {skor}/{total_bobot}')
    if skor == total_bobot:
        print('Sempurna! Lanjut ke project starter.')
    elif skor >= 18:
        print('Bagus! Review soal yang salah, lalu lanjut.')
    else:
        print('Ulangi bagian terkait di lab, coba lagi besok.')

jalankan_kuis()
```
