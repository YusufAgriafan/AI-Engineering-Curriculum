### Soal 1 — Kenapa subword?

Kenapa tokenizer modern (BPE/WordPiece) memecah kata langka menjadi subword,
bukan memakai token `<UNK>` untuk kata yang tidak ada di vocabulary?

- a) Supaya vocabulary lebih kecil sehingga model lebih cepat dilatih
- b) Karena kata langka biasanya tidak penting untuk prediksi
- c) Karakter/subword SELALU tersedia sebagai fallback — model bisa
  men-encode teks APA PUN dan tetap menebak makna dari potongannya
- d) Karena `<UNK>` membuat loss training tidak stabil


### Soal 2 — Kenapa dibagi sqrt(d_k)?

Pada scaled dot-product attention, skor dibagi `sqrt(d_k)`. Yang terjadi jika
TIDAK dibagi:

- a) Skor jadi negatif sehingga softmax tidak valid
- b) Varians skor tumbuh sebesar d_k → softmax mendekati satu-hot → gradien hampir nol
- c) Attention hanya bisa melihat token yang berdekatan
- d) Hasilnya sama saja, pembagian itu hanya konvensi numerik


### Soal 3 — Kenapa decoder butuh causal mask?

Saat melatih GPT-style decoder dengan next-token prediction, causal mask wajib
karena:

- a) Tanpa mask, attention weights bisa bernilai negatif
- b) Tanpa mask, model memakai informasi token MASA DEPAN saat memprediksi
  token sekarang — loss jadi curang dan model tidak bisa generate autoregressif
- c) Mask diperlukan agar jumlah parameter tidak berlebih
- d) Tanpa mask, positional encoding tidak berfungsi


### Soal 4 — Manakah yang murni fungsi posisi?

Properti mana yang BENAR tentang positional encoding sinusoidal?

- a) PE[p] bergantung pada token yang ada di posisi p
- b) PE[p]·PE[q] hanya fungsi dari selisih (p−q) — karena itu relative
  position bisa dibaca dari dot product
- c) Nilai PE membesar tanpa batas saat posisi bertambah
- d) PE dipelajari bersama parameter model, seperti embedding


### Soal 5 — Efek temperature

Saat memanggil LLM API, `temperature` dinaikkan dari 0.2 ke 1.5. Efeknya pada
distribusi sampling:

- a) Distribusi jadi lebih tajam — jawaban lebih deterministik
- b) Distribusi jadi lebih datar — output lebih bervariasi/kreatif, risiko
  tidak masuk akal naik
- c) Tidak mengubah distribusi, hanya mengubah jumlah token output
- d) Token langka otomatis dihapus dari kandidat


### Soal 6 — Encoder vs decoder

Kamu perlu membangun sistem klasifikasi sentimen (label: positif/negatif/netral)
dengan budget kecil. Pilihan arsitektur paling tepat:

- a) Decoder-only besar (GPT-style) — generate jawaban lalu parsing
- b) Encoder-only (BERT-style) — bidirectional, cukup untuk memahami satu
  kalimat, jauh lebih murah di-inferensi
- c) Encoder-decoder (T5) — selalu paling akurat untuk semua task
- d) Bukan model bahasa — pakai CNN untuk teks

## 🧮 Penilaian Otomatis


```python
import numpy as np

total_bobot = 22


def jalankan_kuis():
    skor = 0
    rincian = []

    def cek_pilihan(nomor, kunci, bobot=1):
        nonlocal skor
        benar = JAWABAN.get(nomor) == kunci
        skor += bobot if benar else 0
        rincian.append((nomor, benar))

    def cek_kode(nomor, uji, bobot=1):
        nonlocal skor
        benar = False
        try:
            benar = bool(uji())
        except Exception as e:
            print(f'  soal {nomor}: error -> {type(e).__name__}: {e}')
        skor += bobot if benar else 0
        rincian.append((nomor, benar))

    def approx(a, b, tol=1e-6):
        return np.allclose(np.asarray(a, dtype=float), np.asarray(b, dtype=float), atol=tol)

    def _safe(nama, *args):
        fn = globals().get(nama)
        if fn is None:
            return None
        try:
            return fn(*args)
        except Exception:
            return None

    cek_pilihan(1, 'c')
    cek_pilihan(2, 'b')
    cek_pilihan(3, 'b')
    cek_pilihan(4, 'b')
    cek_pilihan(5, 'b')
    cek_pilihan(6, 'b')

    # Soal 7 - softmax_stabil
    def uji7():
        p = softmax_stabil(np.array([0.0, np.log(3.0)]))
        assert approx(p, [0.25, 0.75]), f'dapat {p}'
        r = np.random.default_rng(0)
        x = r.normal(0, 30, (5, 7))   # skala besar: versi naive akan meledak
        ps = softmax_stabil(x)
        assert np.allclose(ps.sum(axis=-1), 1.0), 'tiap baris jumlah 1'
        assert np.all(ps >= 0) and np.all(np.isfinite(ps)), 'finite & non-negatif'
        baris = ps[0]
        assert np.argmax(baris) == np.argmax(x[0]), 'urutan probabilitas mengikuti skor'
        return True
    cek_kode(7, uji7, bobot=2)

    # Soal 8 - attention
    def uji8():
        np.random.seed(9)
        Q = np.random.randn(4, 8); K = np.random.randn(4, 8); V = np.random.randn(4, 8)
        out, attn = attention(Q, K, V)
        assert out.shape == (4, 8) and attn.shape == (4, 4)
        assert np.allclose(attn.sum(axis=-1), 1.0, atol=1e-9), 'baris softmax jumlah 1'
        assert np.all(attn > 0), 'softmax positif'
        assert np.allclose(out, attn @ V), 'output = attn @ V'
        # skor wajib diskalakan
        e = np.exp(Q @ K.T)
        attn_noscale = e / e.sum(axis=-1, keepdims=True)
        assert not np.allclose(attn, attn_noscale), 'harus dibagi sqrt(d_k)'
        return True
    cek_kode(8, uji8, bobot=5)

    # Soal 9 - apply_causal_mask
    def uji9():
        m = apply_causal_mask(np.array([[1., 2.], [3., 4.]]))
        assert approx(m, [[1, -1e9], [3, 4]]), f'dapat {m}'
        s = np.random.default_rng(3).normal(0, 1, (5, 5))
        mm = apply_causal_mask(s.copy())
        mask = np.triu(np.ones((5, 5)), k=1) == 1
        assert np.allclose(mm[mask], -1e9, atol=100), 'posisi masa depan jadi -1e9'
        assert np.allclose(mm[~mask], s[~mask]), 'elemen lain tetap'
        return True
    cek_kode(9, uji9, bobot=4)

    # Soal 10 - sample_topk
    def uji10():
        lg = np.array([0.1, 2.2, 0.5, 1.3, 0.1, 0.9, 0.2, 0.4])
        r = np.random.default_rng(9)
        for _ in range(50):
            assert sample_topk(lg, 2, r) in (1, 3), 'top-2 hanya dari {1, 3}'
        r2 = np.random.default_rng(0)
        hasil3 = {sample_topk(lg, 3, r2) for _ in range(300)}
        assert hasil3 <= {1, 3, 5}, f'top-3 = {sorted(hasil3)} — harus subset {{1,3,5}}'
        return True
    cek_kode(10, uji10, bobot=5)

    print('=' * 46)
    for nomor, benar in rincian:
        print(f'  Soal {nomor:>2}: {"BENAR" if benar else "salah"}')
    print('=' * 46)
    print(f'SKOR AKHIR: {skor}/{total_bobot}')
    if skor == total_bobot:
        print('Sempurna! Lanjut ke project starter.')
    elif skor >= 17:
        print('Bagus! Review soal yang salah, lalu lanjut.')
    else:
        print('Ulangi bagian terkait di lab, coba lagi besok.')

jalankan_kuis()
```
