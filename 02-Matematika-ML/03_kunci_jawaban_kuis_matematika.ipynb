## Bagian A — Pilihan Ganda

### Soal 1 — Jawaban: **a) `0`**

`1·2 + 2·(-1) + 2·0 = 2 - 2 + 0 = 0`. Vektor `[1,2,2]` dan `[2,-1,0]` **tegak lurus** (orthogonal) — dot product nol berarti tak ada "kesamaan arah" sama sekali.

### Soal 2 — Jawaban: **b) `0` sampai `1`**

Cosine = cos sudut antar vektor. Untuk vektor non-negatif, sudut terkecil 0° (cos=1), terbesar 90° (cos=0) — tak mungkin berlawanan (itu butuh komponen bertanda berlawanan).

**Di RAG:** embedding selalu non-negatif? Tidak selalu, tapi similarity embedding teks praktisnya berkisar 0–1 — makanya threshold seperti 0.75 masuk akal.

### Soal 3 — Jawaban: **a) `1.8`**

`J(w) = (w-5)²` → `J'(w) = 2(w-5)`. Di `w=1`: gradien = `2(1-5) = -8`.

`w_baru = 1 - 0.1 × (-8) = 1 + 0.8 = 1.8`.

**Perhatikan:** gradien negatif → langkah positif → w menuju 5. Itulah kenapa rumusnya minus: `w -= lr × grad`.

### Soal 4 — Jawaban: **c) `B @ v`**

Aturan bentuk: `(m,n) @ (n,k)` → `(m,k)`. Kolom kiri harus = baris kanan.

- a) `(3,4) @ (4,2)` → `(3,2)` ✅
- b) `(3,4) @ (4,)` → `(3,)` ✅
- c) `(4,2) @ (4,)` → kolom 2 ≠ baris 4 ❌
- d) `(2,4) @ (4,2)` → `(2,2)` ✅

## Bagian B — Coding (versi idiomatik)

```python
import numpy as np

# Soal 5 — mean dan range
def mean_range(data):
    if not data:
        return (0, 0)
    return (sum(data) / len(data), max(data) - min(data))

# Kenapa: guard clause dulu; return tuple langsung tanpa variabel perantara.
print(mean_range([3, 7, 5]), mean_range([]))
```

```python
# Soal 6 — dot product manual
def dot_manual(a, b):
    return sum(x * y for x, y in zip(a, b))

# Kenapa: zip menggandengkan elemen; sum generator = pola reduksi standar.
print(dot_manual([1, 2, 2], [2, -1, 0]))
```

```python
# Soal 7 — cosine similarity dengan guard vektor nol
def cos_sim(a, b):
    norm_a = np.linalg.norm(a)
    norm_b = np.linalg.norm(b)
    if norm_a == 0 or norm_b == 0:
        return 0.0
    return float(np.dot(a, b) / (norm_a * norm_b))

# Kenapa: pembagi nol memberi nan + RuntimeWarning — guard clause lebih jelas daripada try/except.
print(cos_sim(np.array([1., 0.]), np.array([2., 0.])),
      cos_sim(np.array([0., 0.]), np.array([1., 1.])))
```

```python
# Soal 8 — satu langkah gradient descent
def langkah_gd(w, grad_fn, lr):
    return w - lr * grad_fn(w)

# Kenapa: inilah seluruh algoritma — update = posisi - langkah ke arah turun.
grad = lambda w: 2 * (w - 5)
print(langkah_gd(1.0, grad, 0.1))
```

```python
# Soal 9 — z-score
def zscore(data):
    data = np.asarray(data, dtype=float)
    return (data - data.mean()) / data.std()

# Kenapa: np.asarray menormalkan input list/tuple; vektorisasi tanpa loop.
print(zscore([10, 20, 30]).round(3))
```

```python
# Soal 10 — cek menurun (datar boleh)
def loss_menurun(daftar_loss):
    return all(daftar_loss[i] >= daftar_loss[i+1] for i in range(len(daftar_loss) - 1))

# Kenapa: all() + generator = pola 'semua pasangan memenuhi syarat'.
# >= bukan > karena datar (konvergen) dianggap OK.
print(loss_menurun([5, 4, 4, 3]), loss_menurun([5, 4, 5]), loss_menurun([7]))
```
