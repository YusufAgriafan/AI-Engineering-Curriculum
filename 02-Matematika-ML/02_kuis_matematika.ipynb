## 🧮 Penilaian Otomatis

```python
total_bobot = 15

def jalankan_kuis():
    skor = 0
    rincian = []

    def cek_pilihan(nomor, kunci, bobot=1):
        nonlocal skor
        benar = JAWABAN.get(nomor) == kunci
        skor += bobot if benar else 0
        rincian.append((nomor, benar))

    def cek_kode(nomor, uji, bobot=1):
        nonlocal skor
        benar = False
        try:
            benar = bool(uji())
        except Exception as e:
            print(f'  soal {nomor}: error -> {type(e).__name__}: {e}')
        skor += bobot if benar else 0
        rincian.append((nomor, benar))

    cek_pilihan(1, 'a')
    cek_pilihan(2, 'b')
    cek_pilihan(3, 'a')
    cek_pilihan(4, 'c')

    cek_kode(5, lambda: mean_range([3, 7, 5]) == (5.0, 4) and mean_range([]) == (0, 0), bobot=2)
    cek_kode(6, lambda: dot_manual([1, 2, 2], [2, -1, 0]) == 0 and dot_manual([1, 1], [1, 1]) == 2)
    cek_kode(7, lambda: abs(cos_sim(np.array([1., 0.]), np.array([0., 1.]))) < 1e-9
                      and abs(cos_sim(np.array([1., 0.]), np.array([2., 0.])) - 1) < 1e-9
                      and cos_sim(np.array([0., 0.]), np.array([1., 1.])) == 0.0, bobot=2)
    cek_kode(8, lambda: abs(langkah_gd(1.0, lambda w: 2 * (w - 5), 0.1) - 1.8) < 1e-9, bobot=2)
    cek_kode(9, lambda: abs(sum(zscore([10, 20, 30]))) < 1e-9
                      and abs(max(zscore([10, 20, 30])) - min(zscore([10, 20, 30])) - 2.449489742783178) < 1e-6, bobot=2)
    cek_kode(10, lambda: loss_menurun([5, 4, 4, 3]) is True and loss_menurun([5, 4, 5]) is False
                       and loss_menurun([7]) is True, bobot=2)

    print('=' * 46)
    for nomor, benar in rincian:
        print(f'  Soal {nomor:>2}: {"BENAR" if benar else "salah"}')
    print('=' * 46)
    print(f'SKOR AKHIR: {skor}/{total_bobot}')
    if skor == total_bobot:
        print('Sempurna! Lanjut ke project starter.')
    elif skor >= 12:
        print('Bagus! Review soal yang salah, lalu lanjut.')
    else:
        print('Ulangi bagian terkait di lab, coba lagi besok.')

jalankan_kuis()
```
