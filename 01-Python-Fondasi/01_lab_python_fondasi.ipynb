## Bagian 2 — Control Flow & Comprehension

```python
# Demo — enumerate & zip: duo yang sering menggantikan loop manual
kata = ['halo', 'dunia', 'ai']
frekuensi = [12, 8, 30]

for i, k in enumerate(kata):
    print(f'{i}: {k}')

for k, f in zip(kata, frekuensi):
    print(f'{k} muncul {f}x')

# Dict comprehension: bangun kamus frekuensi dalam 1 baris
frekuensi_kata = {k: f for k, f in zip(kata, frekuensi)}
print(frekuensi_kata)
```

## Bagian 3 — Fungsi

Fungsi yang baik: punya nama jelas, docstring, type hint, dan **satu tugas**.

```python
# Demo — default, keyword, *args/**kwargs
def normalisasi(angka: list[float], panggil_debug: bool = False, **opsi) -> list[float]:
    """Skalakan list angka ke rentang 0-1 (min-max)."""
    lo, hi = min(angka), max(angka)
    if opsi.get('metode') == 'zscore':
        import statistics
        m = statistics.mean(angka)
        s = statistics.pstdev(angka)
        return [(x - m) / s for x in angka]
    if panggil_debug:
        print(f'debug: min={lo}, max={hi}')
    return [(x - lo) / (hi - lo) for x in angka]

print(normalisasi([10, 20, 30]))
print(normalisasi([10, 20, 30], panggil_debug=True))
print(normalisasi([10, 20, 30], metode='zscore'))
```

## Bagian 4 — OOP

Layer TensorFlow (Bab 4) dan tool agent (Bab 14) adalah class. Pahami `__init__`, `__repr__`, `__call__`.

```python
# Demo — class dengan __init__, method, __repr__, __call__
class TokenCounter:
    def __init__(self, token_per_kata: float = 1.3):
        self.token_per_kata = token_per_kata
        self.total = 0

    def hitung(self, teks: str) -> int:
        n = int(len(teks.split()) * self.token_per_kata)
        self.total += n
        return n

    def __call__(self, teks: str) -> int:      # bikin objek bisa dipanggil seperti fungsi
        return self.hitung(teks)

    def __repr__(self):
        return f'TokenCounter(tpk={self.token_per_kata}, total={self.total})'

tc = TokenCounter()
print(tc('Halo dunia ini contoh teks'))   # __call__
print(tc.hitung('satu dua tiga'))
print(tc)                                  # __repr__
```

## Bagian 5 — Error Handling

Tangkap **spesifik**, bukan `except Exception`. Raise custom exception untuk aturan bisnis.

```python
# Demo — custom exception + try/except/else/finally
class DataKosongError(Exception):
    """Dilempar saat data yang diminta kosong."""

def ambil_batch(batch: list, n: int):
    if not batch:
        raise DataKosongError('batch kosong — cek sumber data dulu')
    try:
        hasil = batch[:n]
    except TypeError as e:
        print('TypeError:', e)
        return None
    else:
        return hasil
    finally:
        print('--- selesai ambil_batch ---')

print(ambil_batch([1, 2, 3], 2))
try:
    ambil_batch([], 2)
except DataKosongError as e:
    print('Tertangkap:', e)
```

## Bagian 6 — NumPy

Aturan emas: **jangan loop Python kalau bisa vectorized**.

```python
import numpy as np
import time

a = np.array([1.0, 2.0, 3.0])
b = np.array([4.0, 5.0, 6.0])
print('dot:', a @ b, '| norm:', np.linalg.norm(a), '| shape:', a.shape)

# Benchmark loop vs vectorized (dot product 1 juta elemen)
n = 1_000_000
x, y = np.random.rand(n), np.random.rand(n)

t0 = time.perf_counter()
total = 0.0
for i in range(n):
    total += x[i] * y[i]
t_loop = time.perf_counter() - t0

t0 = time.perf_counter()
_ = x @ y
t_vec = time.perf_counter() - t0

print(f'loop: {t_loop:.4f}s | numpy: {t_vec:.6f}s | {t_loop/t_vec:.0f}x lebih cepat')

# Matriks fitur untuk latihan berikutnya (baris = rumah, kolom = fitur):
X = np.array([
    [120.0, 3.0,  5.0],
    [ 80.0, 2.0, 12.0],
    [150.0, 4.0,  2.0],
    [ 95.0, 2.0,  8.0],
])
print('X shape:', X.shape)
```
