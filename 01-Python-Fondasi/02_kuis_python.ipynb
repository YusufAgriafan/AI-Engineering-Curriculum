## 🧮 Penilaian Otomatis

````python
total_bobot = 15

def jalankan_kuis():
    skor = 0
    rincian = []

    def cek_pilihan(nomor, kunci, bobot=1):
        nonlocal skor
        benar = JAWABAN.get(nomor) == kunci
        skor += bobot if benar else 0
        rincian.append((nomor, benar))

    def cek_kode(nomor, uji, bobot=1):
        nonlocal skor
        benar = False
        try:
            benar = bool(uji())
        except Exception as e:
            print(f'  soal {nomor}: error -> {type(e).__name__}: {e}')
        skor += bobot if benar else 0
        rincian.append((nomor, benar))

    cek_pilihan(1, 'b')
    cek_pilihan(2, 'c')
    cek_pilihan(3, 'd')
    cek_pilihan(4, 'a')

    cek_kode(5, lambda: lamaran(2) == 1 and lamaran(5) == 1 and lamaran(1) == 0 and lamaran(0) == 0, bobot=2)
    cek_kode(6, lambda: jumlah_vokal('Kucing Oren') == 4 and jumlah_vokal('xyz') == 0, bobot=2)
    cek_kode(7, lambda: rata_rata([2, 4, 6]) == 4.0 and rata_rata([]) == 0, bobot=2)
    cek_kode(8, lambda: B == {'satu': 1, 'dua': 2, 'tiga': 3})
    cek_kode(9, lambda: LabelCleaner().bersihkan('x y') == ['UNKNOWN', 'y'] and LabelCleaner().bersihkan('X z') == ['UNKNOWN', 'z'], bobot=2)
    cek_kode(10, lambda: huruf_terbanyak('hello') == 'l' and huruf_terbanyak('Ab AB a') in ('a', 'b'), bobot=2)

    print('=' * 46)
    for nomor, benar in rincian:
        print(f'  Soal {nomor:>2}: {"BENAR" if benar else "salah"}')
    print('=' * 46)
    print(f'SKOR AKHIR: {skor}/{total_bobot}')
    if skor == total_bobot:
        print('Sempurna! Lanjut ke project starter.')
    elif skor >= 12:
        print('Bagus! Review soal yang salah, lalu lanjut.')
    else:
        print('Ulangi bagian terkait di lab, coba lagi besok.')

jalankan_kuis()
````
