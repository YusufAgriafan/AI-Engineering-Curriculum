## Bagian A — Pilihan Ganda

### Soal 1 — Jawaban: **b) `[1, 2, 3, 4]`**

`b = a` tidak menyalin list — hanya menyalin **referensi**. Keduanya menunjuk ke objek list yang sama, jadi `b.append(4)` mengubah objek yang juga dilihat `a`.

```python
b = a.copy()      # atau a[:] atau list(a) — salinan dangkal (shallow copy)
```

**Terhubung:** kesalahan klasik serupa: `df2 = df` lalu mengubah `df2` ikut mengubah `df`.

### Soal 2 — Jawaban: **c) `THO`**

`"PYTHON"[-4:-1]` — panjang string 6, jadi `-4` merujuk indeks `6-4 = 2` dan `-1` merujuk `6-1 = 5` sebagai batas **eksklusif**. Sama dengan `[2:5]`.

Jangan percaya penjelasan — jalankan kodenya:

````python
print("PYTHON"[-4:-1])   # output: THO

# Indeks positif:  P(0) Y(1) T(2) H(3) O(4) N(5)
# [-4:-1] == [2:5] == T(2), H(3), O(4) -> 'THO'
# Slicing: batas stop TIDAK ikut — berhenti sebelum N.
````

### Soal 3 — Jawaban: **d) `[x for x in range(4)]`**

- a) hasilnya `[0, 2]` — ada filter genap
- b) hasilnya `[1, 2]` — `range(1, 3)` hanya 1..2
- c) hasilnya `[1, 2]` — sama, batas atas range eksklusif
- d) hasilnya `[0, 1, 2, 3]` ✅ — `range(4)` = 0,1,2,3

**Poin:** `range(n)` dimulai dari 0 dan **tidak termasuk** `n`; slice `[a:b]` tidak termasuk `b`.

### Soal 4 — Jawaban: **a) `1`**

`a` = `{'a', 'b', 'c'}` (duplikat `'a'` hilang). `a - b` = elemen di `a` tapi tidak di `b` = `{'b'}` → `len` = 1.

Operasi set yang sering dipakai di data: `-` (selisih), `&` (irisan), `|` (gabungan), `^` (selisih simetris).

## Bagian B — Coding (versi idiomatik)

````python
# Soal 5 — fungsi syarat
def lamaran(tahun_pengalaman):
    return int(tahun_pengalaman >= 2)

# Kenapa: perbandingan SUDAH menghasilkan bool; int() mengubahnya jadi 1/0.
# Alternatif eksplisit (juga benar):
def lamaran_v2(tahun_pengalaman):
    if tahun_pengalaman >= 2:
        return 1
    return 0

print(lamaran(2), lamaran(1), lamaran(5), lamaran(0))
````

````python
# Soal 6 — hitung vokal
def jumlah_vokal(teks):
    return sum(1 for ch in teks.lower() if ch in 'aiueo')

# Kenapa: generator expression + sum = pola 'hitung yang memenuhi syarat' paling ringkas.
# 'in' pada string = pengecekan keanggotaan.
print(jumlah_vokal('Kucing Oren'))  # 4
````

````python
# Soal 7 — rata-rata aman
def rata_rata(angka):
    if not angka:          # pola guard clause: tangani kasus tepi lebih dulu
        return 0
    return sum(angka) / len(angka)

# Kenapa: cek kosong DI AWAL lebih mudah dibaca daripada try/except ZeroDivisionError.
print(rata_rata([2, 4, 6]), rata_rata([]))
````

````python
# Soal 8 — bangun dict
A = {'satu': 1, 'dua': 2}
tiga = 'tiga'

B = {**A, tiga: 3}       # unpacking dict + kunci baru
print(B)

# Alternatif: B = A | {tiga: 3}  (Python 3.9+) — merge operator
# Hindari: B = A.copy(); B[tiga] = 3 — benar tapi lebih verbose
````

````python
# Soal 9 — class sederhana
class LabelCleaner:
    def bersihkan(self, teks):
        return ['UNKNOWN' if kata.lower() == 'x' else kata for kata in teks.split()]

# Kenapa: comprehension di dalam method = transformasi 1 baris; .lower() menormalkan huruf besar/kecil.
print(LabelCleaner().bersihkan('x y'))
print(LabelCleaner().bersihkan('X z'))
````

````python
# Soal 10 — huruf terbanyak
def huruf_terbanyak(teks):
    hitung = {}
    for ch in teks.lower():
        if ch != ' ':
            hitung[ch] = hitung.get(ch, 0) + 1
    return max(hitung, key=hitung.get)

# Kenapa: hitung.get(ch, 0) + 1 = pola counter manual (pahami SEBELUM pakai collections.Counter).
# max(hitung, key=hitung.get) = kunci dengan nilai terbesar.
print(huruf_terbanyak('hello'), huruf_terbanyak('Ab AB a'))

# Versi stdlib untuk pembanding:
from collections import Counter
print(Counter('hello'.replace(' ', '').lower()).most_common(1))
````
